## 1. Ranked actions + reason codes

## Ranked Actions

### Action 1
- **Action:** Review SEO
- **Reason Code:** CTR_POSITION_OPPORTUNITY
- **Why:** High search volume with low CTR and poor average position suggests strong opportunity for SEO improvement.

### Action 2
- **Action:** Improve Metadata
- **Reason Code:** LOW_CTR
- **Why:** Pages receive impressions but attract relatively few clicks.

### Action 3
- **Action:** Improve Rankings
- **Reason Code:** HIGH_POSITION
- **Why:** Pages rank lower in search results and could benefit from optimization.

### Action 4
- **Action:** Monitor Performance
- **Reason Code:** STABLE_PAGE
- **Why:** Pages already perform reasonably well and only require monitoring.

```python
import pandas as pd
import os

# Load dataset
df = pd.read_csv("/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")

# Keep required columns
df = df[["content_id", "search_volume", "avg_position", "ctr"]].dropna()

# Simple action score
df["action_score"] = (
    (df["search_volume"] / df["search_volume"].max()) * 0.5 +
    (1 - df["ctr"] / 100) * 0.3 +
    (df["avg_position"] / df["avg_position"].max()) * 0.2
)

# Reason codes
def reason(row):
    if row["ctr"] < 5:
        return "LOW_CTR"
    elif row["avg_position"] > 20:
        return "HIGH_POSITION"
    else:
        return "CTR_POSITION_OPPORTUNITY"

df["reason_code"] = df.apply(reason, axis=1)
df["action"] = "Review SEO"

# Rank pages
ranked = df.sort_values("action_score", ascending=False).head(20)

print("Top 20 Ranked Actions")
display(
    ranked[[
        "content_id",
        "action",
        "reason_code",
        "action_score"
    ]]
)
```

## 2. Intended use and limits

## Intended Use

This content action playbook is intended for SEO analysts, digital marketers, and website owners. It helps prioritize webpages that should be reviewed first based on observed search performance signals such as CTR, average position, and search volume. The ranked actions are designed to support decision-making and manual SEO review.

## Limits

- The recommendations are based only on the available dataset and selected features.
- The action scores are decision-support signals, not guaranteed improvements.
- Human review is required before making any content or SEO changes.
- The model does not consider content quality, user intent, backlinks, or other external ranking factors.
- The playbook should not be used for fully automated SEO decisions.

## 3. Human review + the no-go list

## Human Review

Before taking action, a human reviewer should check:

- Whether the page content is accurate and up to date.
- Whether the recommended SEO changes match the page's purpose.
- Whether there are business or client-specific reasons to keep the page unchanged.
- Whether recent updates or campaigns already explain the current performance.
- Whether improving the page is likely to provide value compared with the required effort.

## No-Go List

The following actions should **not** be automated:

- Automatically publishing content changes.
- Automatically deleting or redirecting webpages.
- Automatically changing titles, headings, or metadata without review.
- Making business decisions based only on the action score.
- Applying recommendations without human approval.

This playbook is intended for **decision-support**. Final actions should always be reviewed by a human.

```python
print("Human review rules and no-go list documented.")
```

## 4. Monitoring / retrain triggers

## Monitoring / Retrain Triggers

The recommendations should be reviewed and the model should be retrained if any of the following occur:

- Average CTR changes significantly over time.
- Average search position changes for many pages.
- Search volume patterns change because of seasonal or business trends.
- New content is added or a large amount of existing content is updated.
- Model performance becomes worse than the current baseline.
- Regular monitoring shows that the ranked recommendations are no longer useful.

The action playbook should be reviewed periodically and updated whenever new data becomes available. Retraining is intended to maintain decision-support quality rather than fully automate decisions.
