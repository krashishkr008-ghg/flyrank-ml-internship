## Verification Queries

The verification queries confirmed the following:

- The dataset contains **30,000 rows** and **44 columns**.
- The combination of **content_id** and **client_id** has **0 duplicate rows**, supporting the chosen unit of analysis.
- The **search_volume** field has **2,468 missing values**, while **avg_position** and **ctr** have no missing values.
- The starter dataset does not contain a **month** column, so no date window could be verified from this dataset.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv("flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")

print("========== ROW COUNT ==========")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\n========== MISSING VALUES ==========")
cols = ["avg_position", "search_volume", "impressions", "ctr"]
existing = [c for c in cols if c in df.columns]
print(df[existing].isnull().sum())

print("\n========== DATE WINDOW ==========")
if "month" in df.columns:
    print("Min month:", df["month"].min())
    print("Max month:", df["month"].max())
else:
    print("No month column found.")

print("\n========== GRAIN CHECK ==========")
if {"content_id", "client_id"}.issubset(df.columns):
    duplicates = df.duplicated(subset=["content_id", "client_id"]).sum()
    print("Duplicate content_id + client_id rows:", duplicates)
else:
    print("content_id/client_id columns not found.")
```
