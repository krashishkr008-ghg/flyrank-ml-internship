## 2. Build the ranked queue (writes the CSV)

## Baseline Action Score

Pages are scored using three signals:

- Higher search volume → higher priority.
- Lower CTR → higher priority.
- Poor average position → higher priority.

The score is used to rank pages for SEO review.

Reason Code:
CTR_POSITION_OPPORTUNITY

Action Label:
Review SEO

```python
import pandas as pd
import os

# Load dataset
df = pd.read_csv("/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")

# Fill missing search volume
df["search_volume"] = df["search_volume"].fillna(0)

# Calculate baseline score
df["baseline_score"] = (
    df["search_volume"] * 0.5 +
    (100 - df["ctr"]) * 0.3 +
    df["avg_position"] * 0.2
)

# Reason code
df["reason_code"] = "CTR_POSITION_OPPORTUNITY"

# Action label
df["action_label"] = "Review SEO"

# Rank pages
df = df.sort_values("baseline_score", ascending=False)

# Create output folder
os.makedirs("work/outputs", exist_ok=True)

# Save CSV
df.to_csv("work/outputs/baseline_action_score.csv", index=False)

# Display Top 10
print("Top 10 Ranked Pages")
display(df[[
    "content_id",
    "client_id",
    "search_volume",
    "ctr",
    "avg_position",
    "baseline_score",
    "reason_code",
    "action_label"
]].head(10))

print("\nCSV saved to:")
print("work/outputs/baseline_action_score.csv")
```

## 3. Top-20 review

## Top-20 Review

The top 20 ranked pages were reviewed using the baseline action score.

For each page, I recorded:
- Action
- Reason code
- Confidence note
- What would make the recommendation wrong

The review is intended as decision-support rather than a final recommendation.

```python
# Review the Top 20 ranked pages

top20 = df.head(20)

for rank, (_, row) in enumerate(top20.iterrows(), start=1):
    print(f"Rank: {rank}")
    print(f"Content ID: {row['content_id']}")
    print(f"Action: {row['action_label']}")
    print(f"Reason Code: {row['reason_code']}")
    print("Confidence Note: Medium - based on search volume, CTR, and average position only.")
    print("What would make it wrong: Search intent may have changed, data may be outdated, or the page may already have been optimized.")
    print("-" * 80)
```

## 4. ## Weak Picks + Leakage Check

### Weak Picks

Some pages may receive a high baseline score because of temporary changes in search volume or CTR. The baseline rule does not consider seasonality, recent content updates, or user intent. Therefore, some recommendations may require manual review before action.

### Leakage Check

Observed: No product flags, future-window variables, or label-derived fields were used to calculate the baseline score.

The baseline score uses only:
- Search Volume
- CTR
- Average Position

These signals are intended for decision-support and do not include future information or product decision flags.

```python
# Weak picks review

print("Weak Picks")
print("-" * 60)

print("The baseline rule may incorrectly prioritize:")
print("• Seasonal pages with temporarily low CTR.")
print("• Pages whose search volume has recently changed.")
print("• Pages that were recently updated but the dataset does not reflect those updates.")

print("\nLeakage Check")
print("-" * 60)

used_features = ["search_volume", "ctr", "avg_position"]

print("Features used:", used_features)
print("No label-derived fields or future-window fields were used.")
print("No product flags were included in the baseline score.")
```
