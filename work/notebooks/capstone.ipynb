## 3. Methodology

## Assumptions
- Search volume and average position are useful indicators of search performance.
- CTR is treated as the outcome to be predicted.
- The model is intended for decision-support and not for automatic content changes.

## Features
The model uses the following input features:
- Search Volume
- Average Position

## Label Definition
The prediction target (label) is Click-Through Rate (CTR).

## Baseline
The Week-4 baseline used a simple action score based on search volume, average position, and CTR to rank pages for SEO review.

## Validation Design
A grouped train-test split based on Client ID was used. This prevents pages from the same client appearing in both the training and testing sets, providing a more honest estimate of model performance.

## Leakage Checks
Only search signals available before prediction were used as model features. Client ID was used only for grouped validation and was never used as an input feature. No future information, product flags, or private client data were included.

```python
print("Methodology Summary")
print("-------------------")
print("Features           : search_volume, avg_position")
print("Label              : ctr")
print("Baseline           : Week-4 Action Score")
print("Validation Design  : Grouped Split using client_id")
print("Leakage Check      : Passed")
```
